```python
null_value=df['Logistics_Delay_Reason'].mode()[0]
df['Logistics_Delay_Reason']=df['Logistics_Delay_Reason'].fillna(null_value)
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns
```

```python
def asset_utilization_category(x):
    if 60<= x <=70:
        return "Low Utilization"
    elif 71<= x <= 85:
        return "Medium Utilization"
    elif 86<= x <=100:
        return "High Utilization"
    else:
        return "Outside Range"

df['asset_utilization_category']=df['asset_utilization'].apply(asset_utilization_category)

```

```python
df[['asset_utilization','asset_utilization_category']]
```

```python
df['timestamp']=pd.to_datetime(df['timestamp'])
```

```python
df['month_name']=df['timestamp'].dt.month_name()
df['month']=df['timestamp'].dt.month
df['week']=df['timestamp'].dt.isocalendar().week
df['day']=df['timestamp'].dt.day
df['day_of_week']=df['timestamp'].dt.day_name()
df['hour']=df['timestamp'].dt.hour
df['date']=df['timestamp'].dt.date

df[['month_name','month','week','day','day_of_week','hour','date']]
```

```python
def purchase_frequency_category(x):
    if 1<=x<=3:
        return "Low"
    elif 4<=x<=7:
        return "Medium"
    else:
        return "High"

df['purchase_frequency_category']=df['user_purchase_frequency'].apply(purchase_frequency_category)
df[['purchase_frequency_category','user_purchase_frequency']]
```

```python
#Exporting the cleaned data:
df.to_csv('smart_logistics_dataset.csv', index= False)
```
